```python
import glob
review_files = glob.glob('sephora_data/reviews_*.csv')
review_files
```

```python
all_reviews = pd.concat([pd.read_csv(file, low_memory=False) for file in review_files], ignore_index=True)
all_reviews.shape
```

```python
dupe_mentions_full = all_reviews[all_reviews['review_text'].str.contains('dupe', case=False, na=False)]
dupe_mentions_full.shape
```

```python
dupe_full_with_category = dupe_mentions_full.merge(products[['product_id', 'primary_category']], on='product_id', how='left')
dupe_full_with_category['primary_category'].value_counts(dropna=False)
```

```python
dupe_mentions_full = all_reviews[all_reviews['review_text'].str.contains('dupe', case=False, na=False)]
dupe_mentions_full.shape
```

```python
products['regulation_flag'] = products['ingredients'].str.contains('retinol|glycolic|salicylic|lactic acid', case=False, na=False)
products.groupby('primary_category')['regulation_flag'].sum().sort_values(ascending=False)
```

```python
dupe_by_category = dupe_full_with_category['primary_category'].value_counts()
reg_by_category = products.groupby('primary_category')['regulation_flag'].sum()

threat_summary = pd.DataFrame({
    'dupe_mentions': dupe_by_category,
    'regulation_flagged_products': reg_by_category
}).fillna(0)

threat_summary
```

```python
threat_summary['dupe_score'] = (threat_summary['dupe_mentions'] / threat_summary['dupe_mentions'].max()) * 100
threat_summary['regulation_score'] = (threat_summary['regulation_flagged_products'] / threat_summary['regulation_flagged_products'].max()) * 100
threat_summary['combined_threat_score'] = (threat_summary['dupe_score'] + threat_summary['regulation_score']) / 2

threat_summary.sort_values('combined_threat_score', ascending=False)
```

```python
products.to_csv('products_clean.csv', index=False)
threat_summary.to_csv('threat_summary.csv', index=False)
```
